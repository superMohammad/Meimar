```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect('data/archive/database.db')
```

```python
chunks = pd.read_sql(
    "SELECT * FROM listings",
    conn,
    chunksize=100_000
)

for i, chunk in enumerate(chunks):
    chunk.to_parquet(
        f"part_{i}.parquet",
        engine="pyarrow",
        compression="snappy",
        index=False
    )
```
